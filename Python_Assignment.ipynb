**Topic 7: Functions & Passing Arguments**

Q7.1. Write a function my_abs(val) that returns the absolute value of val without using Python's built-in abs().
Test it with a positive number, a negative number, and 0.

```python
def my_abs(val):
    if val < 0:
        return -val
    else:
        return val
# Testing the function
print(my_abs(10))
print(my_abs(-10))
print(my_abs(0))
```

```python
counter = 0
def increment_counter():
    counter = counter + 1
increment_counter()
#Because we are assigning a value to counter, Python treats counter as a local variable.
#But on the right side, counter + 1, Python tries to read this local counter before it has been given a value.
#Therefore, Python gives an UnboundLocalError.
```

Q11. Class Attendance Counter
• Create a global variable total_students = 0.
• Write a function mark_present() that increases total_students by 1 — remember
to use the global keyword inside the function so it can change the global variable.
• Call mark_present() three times (once for each student who walks in), then print

```python
total_students = 0
def mark_present():
    global total_students
    total_students += 1
mark_present()
mark_present()
mark_present()
print(total_students)
```
